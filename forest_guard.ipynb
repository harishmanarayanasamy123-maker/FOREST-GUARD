```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Check if the file exists, if not, create dummy sensor data for testing
csv_filename = "iot_sensor_data.csv"
if not os.path.exists(csv_filename):
    print(f"'{csv_filename}' not found. Generating mock sensor data for you...")
    mock_data = {
        "temp_c": [25.3, 28.1, 30.5, 41.2, 35.0, 39.8, 42.1, 31.4, 29.9, 38.5]
    }
    pd.DataFrame(mock_data).to_csv(csv_filename, index=False)

# Read your actual sensor data
df = pd.read_csv(csv_filename)

print("First 5 sensor readings:")
print(df.head())

print("\nDataset information:")
print(df.describe())

# Plot temperature readings
plt.figure(figsize=(10, 5))
plt.plot(df["temp_c"], marker="o")
plt.title("Forest Monitoring - Temperature")
plt.xlabel("Reading number")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.tight_layout()
plt.show()

# Basic warning flag for investigation only
df["temperature_warning"] = df["temp_c"] >= 40

print("\nTemperature warning records:")
print(df[df["temperature_warning"]])
```
